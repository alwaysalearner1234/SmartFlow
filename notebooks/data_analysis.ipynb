# 📊 Market Microstructure & Order-Book Data Analysis
This notebook explores multi-level order-book state representations, depth ladders, bid-ask spreads, and trade print flow across different market regimes.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from data.generator import MarketDataGenerator
from data.validator import MarketDataValidator
```

```python
# Generate synthetic snapshot stream
generator = MarketDataGenerator()
snapshots = generator.generate_scenario_stream('normal_market', num_ticks=400)
print(f'Generated {len(snapshots)} snapshots.')
print(f'Arrival Mid-Price: ${snapshots[0].mid_price:.2f}')
print(f'Top-level Spread: ${snapshots[0].spread:.2f} ({snapshots[0].spread / snapshots[0].mid_price * 10000:.1f} bps)')
```

```python
# Convert to DataFrame and inspect
prices = [s.mid_price for s in snapshots]
spreads = [s.spread for s in snapshots]
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(prices, color='#00F0FF')
plt.title('Mid Price Evolution')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(spreads, bins=20, color='#FF0055', alpha=0.7)
plt.title('Spread Distribution')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
