# 🏆 Execution Strategy Backtesting & Benchmark Analysis
Compares Market, TWAP, VWAP, Almgren-Chriss, and the Proposed Risk-Aware SOR strategy under identical market conditions.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
from data.generator import MarketDataGenerator
from backtest.backtester import Backtester
from backtest.comparison import ScenarioComparisonRunner
```

```python
generator = MarketDataGenerator()
snaps = generator.generate_scenario_stream('poor_liquidity', num_ticks=250)
backtester = Backtester()
res = backtester.run_backtest(snaps, scenario_name='Poor Liquidity', total_quantity=1000.0, horizon_sec=60.0)
print(res.comparison_table.to_string(index=False))
```

```python
# Trajectory Comparison
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
for name, strat_res in res.strategy_results.items():
    if strat_res.trajectory:
        times = [pt['elapsed_time'] for pt in strat_res.trajectory]
        rem = [pt['remaining_quantity'] for pt in strat_res.trajectory]
        plt.plot(times, rem, label=name, marker='o', markersize=3)
plt.xlabel('Elapsed Time (s)')
plt.ylabel('Remaining Inventory')
plt.title('Execution Trajectory Comparison')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
