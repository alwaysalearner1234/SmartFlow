# 🔬 Microstructure Feature Engineering & Signal Analysis
Analyzes Order Flow Imbalance (OFI), micro-price divergence, spread expansion, rolling volatility, and trade volume imbalance.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
from data.generator import MarketDataGenerator
from features import build_feature_pipeline
```

```python
generator = MarketDataGenerator()
snaps = generator.generate_scenario_stream('high_adverse_selection', num_ticks=300)
df_feat = build_feature_pipeline(snaps)
print('Extracted features shape:', df_feat.shape)
df_feat[['spread_bps', 'depth_imbalance_l1', 'ofi_instant', 'volatility_std_10']].describe()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
axes[0].plot(df_feat['mid_price'], color='#F3BA2F', label='Mid Price')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(df_feat['depth_imbalance_l1'], color='#00F0FF', label='L1 Depth Imbalance')
axes[1].axhline(0, color='gray', linestyle='--')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

axes[2].bar(df_feat.index, df_feat['ofi_instant'], color='#00FFA3', alpha=0.5, label='OFI')
axes[2].legend(); axes[2].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
